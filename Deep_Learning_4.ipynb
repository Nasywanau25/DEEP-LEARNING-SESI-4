```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np

# Contoh data teks
text1 = ["apple", "banana", "orange", "banana", "apple", "orange", "banana"]
text2 = ["apple", "banana", "orange", "banana", "apple", "orange", "banana", "melon", "apple", "banana", "orange", "melon"]

# Reshape data karena OneHotEncoder mengharapkan input dalam bentuk kolom
words1 = np.array(text1).reshape(-1, 1)
words2 = np.array(text2).reshape(-1, 1)

# Inisialisasi OneHotEncoder (satu encoder untuk tiap dataset)
encoder1 = OneHotEncoder(sparse_output=False)
encoder2 = OneHotEncoder(sparse_output=False)

# Melakukan fit dan transform pada data
one_hot_encoded1 = encoder1.fit_transform(words1)
one_hot_encoded2 = encoder2.fit_transform(words2)

# Menampilkan hasil one-hot encoding
print("===== Dataset 1 =====")
print("Kata-kata:", encoder1.categories_)
print("One-hot vectors:")
print(one_hot_encoded1)

print("\n===== Dataset 2 =====")
print("Kata-kata:", encoder2.categories_)
print("One-hot vectors:")
print(one_hot_encoded2)
```

```python
import numpy as np

# Definisikan input data (sekuens) dan ukuran
inputs = np.array([0.5, 0.8])  # Contoh input dengan dua timestep
timesteps = len(inputs)

# Inisialisasi bobot RNN (u, v, w)
u = np.array([0.9])  # Bobot input ke hidden state
w = np.array([0.5])  # Bobot hidden state sebelumnya ke hidden state sekarang
v = np.array([1.0])  # Bobot hidden state ke output

# Inisialisasi hidden state awal
h_t = 0

# Placeholder untuk menyimpan output
outputs = []

print("Timestep | Input (x_t) | Hidden State (h_t) | Output (y_t)")
print("--------------------------------------------------------")

# Loop melalui setiap timestep untuk menghitung hidden state dan output
for t in range(timesteps):
    x_t = inputs[t]

    # Hitung hidden state pada timestep t
    h_t = np.tanh(u * x_t + w * h_t)

    # Hitung output pada timestep t
    y_t = v * h_t

    # Simpan output
    outputs.append(y_t)

    # Tampilkan hasil per timestep
    print(f"   {t+1}    |    {x_t:.2f}    |     {float(h_t[0]):.2f}      |    {float(y_t[0]):.2f}")

# Konversi outputs menjadi numpy array
outputs = np.array(outputs)
```

```python
import numpy as np
np.random.seed(42)  # supaya hasil sama setiap dijalankan (boleh dihapus)

# Dataset teks sederhana
text1 = ["apple", "banana", "orange", "banana", "apple", "orange", "banana"]
text2 = ["apple", "banana", "orange", "banana", "apple", "orange", "banana", "melon", "apple", "banana", "orange", "melon"]

# Bangun kosakata unik
vocab1 = sorted(set(text1))
vocab_size1 = len(vocab1)

vocab2 = sorted(set(text2))
vocab_size2 = len(vocab2)

# Buat mapping kata ke indeks dan sebaliknya
word_to_index1 = {word: i for i, word in enumerate(vocab1)}
index_to_word1 = {i: word for i, word in enumerate(vocab1)}

word_to_index2 = {word: i for i, word in enumerate(vocab2)}
index_to_word2 = {i: word for i, word in enumerate(vocab2)}

# One-hot encode data teks
def one_hot_encode(word, vocab_size, word_to_index):
    vector = np.zeros(vocab_size)
    vector[word_to_index[word]] = 1
    return vector

# One-hot encoding dataset
inputs1 = np.array([one_hot_encode(word, vocab_size1, word_to_index1) for word in text1])
inputs2 = np.array([one_hot_encode(word, vocab_size2, word_to_index2) for word in text2])

# Inisialisasi bobot RNN (u, v, w)
# Dataset 1
u1 = np.random.rand(vocab_size1, vocab_size1)  # Bobot input ke hidden state
w1 = np.random.rand(vocab_size1, vocab_size1)  # Bobot hidden state sebelumnya ke hidden state sekarang
v1 = np.random.rand(vocab_size1, vocab_size1)  # Bobot hidden state ke output

# Dataset 2
u2 = np.random.rand(vocab_size2, vocab_size2)
w2 = np.random.rand(vocab_size2, vocab_size2)
v2 = np.random.rand(vocab_size2, vocab_size2)

# Inisialisasi hidden state awal
h_t1 = np.zeros(vocab_size1)
h_t2 = np.zeros(vocab_size2)

# Placeholder untuk menyimpan output dan loss
outputs1 = []
losses1 = []
outputs2 = []
losses2 = []

# Fungsi aktivasi tanh dan loss MSE
def tanh(x):
    return np.tanh(x)

def mse_loss(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

# ===================== DATASET 1 =====================
print("===== Dataset 1 =====")
print("Timestep | Input Word | Hidden State (h_t) | Output (tanh) | Predicted Word | Loss")
print("-------------------------------------------------------------------------------")

# Loop melalui setiap timestep untuk menghitung hidden state, output, dan loss
for t in range(len(inputs1) - 1):
    x_t = inputs1[t]
    y_true = inputs1[t + 1]  # Target output adalah kata berikutnya

    # Hitung hidden state pada timestep t menggunakan tanh
    h_t1 = tanh(np.dot(u1, x_t) + np.dot(w1, h_t1))

    # Hitung output menggunakan tanh
    z = np.dot(v1, h_t1)
    output_tanh1 = tanh(z)

    # Hitung loss menggunakan MSE
    loss = mse_loss(y_true, output_tanh1)
    losses1.append(loss)

    # Prediksi kata dengan mencari argmax dari output_tanh
    predicted_index = np.argmax(output_tanh1)
    predicted_word = index_to_word1[predicted_index]

    # Simpan output
    outputs1.append(predicted_word)

    # Tampilkan hasil per timestep
    print(f"   {t+1}    |    {text1[t]}    |     {h_t1}      |    {output_tanh1}    |    {predicted_word}     |    {loss:.4f}")

# Menampilkan kata-kata hasil prediksi
print("\nHasil prediksi kata-kata:")
print(outputs1)

# Menampilkan nilai rata-rata loss
average_loss1 = np.mean(losses1)
print(f"\nRata-rata Loss: {average_loss1:.4f}")

# ===================== DATASET 2 =====================
print("\n\n===== Dataset 2 =====")
print("Timestep | Input Word | Hidden State (h_t) | Output (tanh) | Predicted Word | Loss")
print("-------------------------------------------------------------------------------")

for t in range(len(inputs2) - 1):
    x_t = inputs2[t]
    y_true = inputs2[t + 1]

    h_t2 = tanh(np.dot(u2, x_t) + np.dot(w2, h_t2))

    z = np.dot(v2, h_t2)
    output_tanh2 = tanh(z)

    loss = mse_loss(y_true, output_tanh2)
    losses2.append(loss)

    predicted_index = np.argmax(output_tanh2)
    predicted_word = index_to_word2[predicted_index]

    outputs2.append(predicted_word)

    print(f"   {t+1}    |    {text2[t]}    |     {h_t2}      |    {output_tanh2}    |    {predicted_word}     |    {loss:.4f}")

print("\nHasil prediksi kata-kata:")
print(outputs2)

average_loss2 = np.mean(losses2)
print(f"\nRata-rata Loss: {average_loss2:.4f}")
```
